# 09 — Salary Model Prediction Test

This notebook performs a simple functional check of the salary prediction pipeline.  
Its purpose is purely to verify that the saved salary model and the `salary_predictor` helper in `src/job_intel/models/` work as expected.

### What this notebook does
- Loads the final trained salary model and PCA transformer.
- Uses the `salary_predictor` utility to generate predictions on sample rows.
- Confirms that:
  - the model loads correctly,
  - feature ordering is consistent,
  - categorical encodings map properly,
  - and predictions return without errors.

### Outcome
A basic correctness test ensuring that the production prediction function operates smoothly.  
No modelling decisions or downstream results are generated in this notebook.

## Set up and load data and models

```python
import sys
from pathlib import Path

project_root = Path().resolve().parent.parent
sys.path.append(str(project_root))
project_root
```

```python
import pandas as pd

from src.job_intel.models.salary_predictor import predict_salary
from src.job_intel.config import CH1_PROCESSED_SALARY_MODEL_PCA_DF

df = pd.read_csv(CH1_PROCESSED_SALARY_MODEL_PCA_DF)

df.shape, df.head()
```

## Test 1 - Predict salary for a fre real jobs

```python
test_rows = df.sample(200, random_state=42)

predictions = []
for idx, row in test_rows.iterrows():
    record = row.to_frame().T
    pred = predict_salary(record)
    predictions.append((idx, pred))

predictions
```

```python
pred = pd.DataFrame(predictions).set_index(0)
pred_comp = pd.DataFrame({'observed' : test_rows['sal_mean'], 'predicted' : pred.loc[:,1]})
pred_comp['diff'] = pred_comp['observed'] - pred_comp['predicted']

print(f" Mean diff {pred_comp['diff'].abs().mean()}"), pred_comp
```
