# 12 — Skill Probability Matrix Creation & Evaluation

This notebook generates and inspects the **Skill Probability Matrix**, a core output of Chapter 1.  
Using the fitted skill-requirement models (trained in Notebook 10), it computes for every job posting the **predicted probability that each of the 27 skill groups is required**.

---

## **Objectives**
- Load the final salary-chapter dataset.  
- Apply consistent categorical encoding to match training conditions.  
- Use `build_skill_probability_matrix()` to produce a job-by-skill probability table.  
- Save the resulting matrix for downstream components of the Job Intelligence Engine.  
- Perform exploratory evaluation and sanity checks via plots and distributions.

---

## **Workflow**
1. **Prepare modelling dataset**  
   Load the processed Chapter 1 dataset and ensure all categorical columns match the dtypes used during model training.

2. **Generate Skill Probability Matrix**  
   For each of the 27 skill groups:
   - Load its tuned LightGBM model.  
   - Select all predictor features.  
   - Run `predict_proba` to obtain the probability of skill presence.  
   - Store results in a unified probability matrix (`n_jobs × 27`).

3. **Explore and validate outputs**  
   - Plot the probability distributions for each skill.  
   - Inspect rare, common, and ambiguous skills.  
   - Confirm that predictions align with expectations from variable importance and prevalence statistics.

---

## **Output**
- A saved, structured skill-probability matrix (CSV) used in:
  - Job–skill similarity modelling  
  - Skill ecosystem graph construction  
  - User–job matching  
  - Career transition simulations  
  - Competitiveness and rarity modelling  

---

This notebook finalises the deliverables of Chapter 1’s skill-requirement modelling, producing the probability layer that supports multiple system components in later chapters.
